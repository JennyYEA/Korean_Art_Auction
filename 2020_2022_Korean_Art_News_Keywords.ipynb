```python
culnews_2023 = pd.read_csv('문화 및 문화누리 뉴스 키워드 추출 정보_2023.csv')
culnews_2023.head(3)
```

```python
culnews_2023.info()
```

```python
culnews = culnews_2023.sort_values(by='REGIST_DT')
print(culnews)
```

```python
# 뉴스 키워드 데이터에서 작가 정보 추출하기
keyword = "이우환"
matching_rows = culnews[culnews.apply(lambda row:\
                                      row.astype(str).str.contains("이우환").any(), axis=1)]

print(matching_rows)
```
