```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
youtube = pd.read_excel('작가 별 Youtube 통계 (4).xlsx')
youtube.head(3)
```

```python
print(youtube['view'].mean())
print(youtube['view'].min())
print(youtube['view'].max())
print(youtube['view'].median())
```

```python
youtube['value'] = np.where(youtube['view'] >= 80000, '80k',
                   np.where(youtube['view'] >= 15188, '15k', 'low'))
youtube.head()
```

```python
sns.lineplot(data = youtube, x = 'year', y = 'view', hue = 'artist')
plt.rcParams.update({'figure.figsize' : [8,6]})
```

```python
sns.boxplot(data = youtube, x = 'artist', y = 'view')
plt.rcParams.update({'figure.figsize' : [8,6]})
```
